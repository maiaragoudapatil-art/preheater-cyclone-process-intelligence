```python
low = (
    sensor_data["Cyclone_Inlet_Gas_Temp"] < 700
) & df["time"].notna()

run_id = low.ne(low.shift()).cumsum()

runs = (
    df.loc[low]
      .assign(run_id=run_id[low].values)
      .groupby("run_id")
      .agg(
          start=("time", "min"),
          end=("time", "max"),
          rows=("time", "size")
      )
)

runs["duration"] = runs["end"] - runs["start"]

print(runs.sort_values("duration", ascending=False).head(15))
```

```python
# Take the largest 5 low-temperature runs
top_runs = runs.sort_values("duration", ascending=False).head(5)
for run_id in top_runs.index:
    start = top_runs.loc[run_id, "start"]
    end = top_runs.loc[run_id, "end"]

    period = sensor_data[
        (df["time"] >= start) &
        (df["time"] <= end)
    ]

    print("\n" + "=" * 70)
    print(f"Run {run_id}: {start} -> {end}")
    print("=" * 70)

    print(
        period[
            [
                "Cyclone_Inlet_Gas_Temp",
                "Cyclone_Material_Temp",
                "Cyclone_Gas_Outlet_Temp",
                "Cyclone_Outlet_Gas_draft",
                "Cyclone_cone_draft",
                "Cyclone_Inlet_Draft"
            ]
        ].describe().loc[["mean", "min", "50%", "max"]]
    )
```

```python
material = sensor_data["Cyclone_Material_Temp"]

for value in [0, 900]:
    mask = material == value

    print("\n" + "=" * 60)
    print(f"Material Temperature = {value}")
    print("=" * 60)

    print("Total rows:", mask.sum())

    valid = df.loc[mask & df["time"].notna(), "time"]

    print("Rows with valid timestamps:", len(valid))

    if len(valid) > 0:
        print("First:", valid.min())
        print("Last :", valid.max())

    print("\nOther sensor medians:")
    print(
        sensor_data.loc[mask, [
            "Cyclone_Inlet_Gas_Temp",
            "Cyclone_Gas_Outlet_Temp",
            "Cyclone_Outlet_Gas_draft",
            "Cyclone_cone_draft",
            "Cyclone_Inlet_Draft"
        ]].median()
    )
```

```python
raw_df = pd.read_csv("cyclone_data.csv")

raw_df["time_parsed"] = pd.to_datetime(
    raw_df["time"],
    format="mixed",
    errors="coerce"
)
print(raw_df[["time", "time_parsed"]].head())
print(raw_df[["time", "time_parsed"]].iloc[3450:3465])
print(raw_df["time_parsed"].isna().sum())
print(raw_df["time_parsed"].min())
print(raw_df["time_parsed"].max())
```

```python
df = raw_df.copy()
df["time"] = df["time_parsed"]
df = df.sort_values("time").reset_index(drop=True)
print(df["time"].isna().sum())
print(df["time"].min())
print(df["time"].max())
```

```python
low = (
    sensor_data["Cyclone_Inlet_Gas_Temp"] < 700
)

run_id = low.ne(low.shift()).cumsum()

runs = (
    df.loc[low]
      .assign(run_id=run_id[low].values)
      .groupby("run_id")
      .agg(
          start=("time", "min"),
          end=("time", "max"),
          rows=("time", "size")
      )
)

runs["duration"] = runs["end"] - runs["start"]

print(
    runs
    .sort_values("duration", ascending=False)
    .head(20)
)
```
